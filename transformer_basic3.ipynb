```python
frequencies_to_compare = [3, 2, 1]
num_rows = len(frequencies_to_compare)
num_cols = 2

x_range = 20
num_points = 1000
x = np.linspace(0, x_range, num_points)

subplot_titles = []
for freq in frequencies_to_compare:
    subplot_titles.append(f'Sine Wave (Freq = {freq})')
    subplot_titles.append(f'Cosine Wave (Freq = {freq})')

# Create subplots
fig = make_subplots(
    rows=num_rows,
    cols=num_cols,
    subplot_titles=subplot_titles,
    vertical_spacing=0.08,
    horizontal_spacing=0.1
)

# Define colors for each frequency
colors = ['#636EFA', '#EF553B', '#00CC96', '#AB63FA', '#FFA15A']

# Generate and add sine and cosine waves for each frequency
for i, freq in enumerate(frequencies_to_compare):
    # Sine wave data (left column)
    y_sin = np.sin(2 * np.pi * freq * (x / x_range))
    fig.add_trace(
        go.Scatter(
            x=x,
            y=y_sin,
            mode='lines',
            name=f'Sine Freq = {freq}',
            line=dict(color=colors[i], width=2),
            showlegend=False
        ),
        row=i + 1, col=1
    )

    # Cosine wave data (right column)
    y_cos = np.cos(2 * np.pi * freq * (x / x_range))
    fig.add_trace(
        go.Scatter(
            x=x,
            y=y_cos,
            mode='lines',
            name=f'Cosine Freq = {freq}',
            line=dict(color=colors[i], width=2),
            showlegend=False
        ),
        row=i + 1, col=2
    )

fig.update_layout(
    height=300 * num_rows,
    title_text="Comparison of Sine and Cosine Waves with Different Frequencies",
    title_font_size=20,
    showlegend=False
)

for i in range(num_rows):
    # Left column (Sine)
    fig.update_xaxes(
        title_text="Token Position",
        range=[0, x_range],
        gridcolor='lightgray',
        row=i + 1,
        col=1
    )
    fig.update_yaxes(
        title_text="Amplitude",
        range=[-1.2, 1.2],
        gridcolor='lightgray',
        row=i + 1,
        col=1
    )

    # Right column (Cosine)
    fig.update_xaxes(
        title_text="Position",
        range=[0, x_range],
        gridcolor='lightgray',
        row=i + 1,
        col=2
    )
    fig.update_yaxes(
        title_text="Amplitude",
        range=[-1.2, 1.2],
        gridcolor='lightgray',
        row=i + 1,
        col=2
    )

fig.write_html('sine_cosine_waves_comparison.html')
fig.show()
```

```python
# ฟังก์ชันสร้าง Positional Encoding (Vaswani et al., 2017)
# y = amplitude * np.sin(2 * np.pi * frequency * t)

# position ทำหน้าที่เหมือน t (เวลาหรือ index token)
# angle_rates ทำหน้าที่เหมือนความถี่ของคลื่นในแต่ละ dim

def positional_encoding(position: int, d_model: int) -> np.ndarray:
    dims = np.arange(d_model)
    angle_rates = 1.0 / np.power(10000.0, (dims // 2) * 2 / d_model)
    angles = position * angle_rates
    pe = np.zeros(d_model, dtype=np.float64)
    pe[0::2] = np.sin(angles[0::2])   # even index → sin
    pe[1::2] = np.cos(angles[1::2])   # odd index  → cos
    return pe

def positional_encoding_matrix(num_tokens: int, d_model: int) -> np.ndarray:
    return np.stack([positional_encoding(i, d_model) for i in range(num_tokens)], axis=0)
```

```python
def plot_positional_heatmap_plotly(num_tokens=50, d_model=64):
    pe_mat = positional_encoding_matrix(num_tokens, d_model)
    fig = px.imshow(
        pe_mat,
        color_continuous_scale="RdBu",
        labels=dict(x="Dimension", y="Token Position", color="Value (sin/cos)"),
        title=f"Positional Encoding Heatmap (Tokens 0–{num_tokens-1}, {d_model} dims)"
    )
    fig.update_layout(height=500, width=900)
    fig.show()

plot_positional_heatmap_plotly(num_tokens=20, d_model=10)
```

```python
max_sequence_length = 75
embedding_dimension = 1024

pe_mat = positional_encoding_matrix(max_sequence_length, embedding_dimension)
pe_mat.shape
```

```python
pca = PCA(n_components=3)

positional_encoding_pca = pca.fit_transform(pe_mat)

print(positional_encoding_pca.shape)
print(positional_encoding_pca[:5])
print("\nExplained variance ratio by the first three components", pca.explained_variance_ratio_)
```

```python
fig_pca_scatter_3d = go.Figure(data=go.Scatter3d(
    x=positional_encoding_pca[:, 0],
    y=positional_encoding_pca[:, 1],
    z=positional_encoding_pca[:, 2],
    mode='markers+text',
    text=[str(pos) for pos in range(max_sequence_length)], # Add text labels for each position
    textposition="bottom center",
    marker=dict(
        size=5,
        color=np.arange(max_sequence_length), # Color points by position
        colorscale='Viridis',
        opacity=0.8
    )
))

fig_pca_scatter_3d.update_layout(
    title='PCA Projection of Positional Encoding (3D)',
    scene = dict(
        xaxis_title='Principal Component 1',
        yaxis_title='Principal Component 2',
        zaxis_title='Principal Component 3'
    ),
    margin=dict(l=0, r=0, b=0, t=40),
    height=700
)

fig_pca_scatter_3d.show()
```

```python
token_embedding_layer = model.model.encoder.embed_tokens
token_embedding_layer.weight.shape
```

```python
long_sentence = "In the vast realm of natural language processing, understanding the nuances of how models handle sequential data is crucial. Positional encoding plays a vital role in providing this essential information to the model, allowing it to differentiate between words at different positions in a sentence, which is fundamental for tasks like translation, summarization, and text generation."
```

```python
tokenizer = AutoTokenizer.from_pretrained(checkpoint)
```

```python
tokens = tokenizer(long_sentence, return_tensors="pt")

print(tokens['input_ids'][0])
```

```python
len(tokens['input_ids'][0])
```

```python
token_embeddings = token_embedding_layer(tokens['input_ids'][0])

print("Token Embedding Matrix shape", token_embeddings.shape)
token_embeddings
```

```python
token_embeddings_np = token_embeddings.detach().numpy()

fig = px.imshow(
    token_embeddings_np,
    color_continuous_scale="RdBu",
    labels=dict(x="Embedding Dimension", y="Token Index", color="Value"),
    title="Token Embedding Heatmap"
)

fig.update_xaxes(side="top")
fig.update_layout(height=500, width=900)
fig.show()
```

```python
te_values = token_embeddings_np.flatten()

fig_hist = px.histogram(te_values, nbins=100, title='Distribution of Token Embedding Values')

fig_hist.update_layout(
    xaxis_title='Value',
    yaxis_title='Count',
    bargap=0.05,
    showlegend=False
)

fig_hist.show()
```

```python
print("Token embedding matrix")
print("Min", np.min(token_embeddings_np))
print("Max", np.max(token_embeddings_np))
print("Average", np.average(token_embeddings_np))
```

```python
token_embeddings_with_pe = token_embeddings_np + pe_mat

token_embeddings_with_pe.shape
```

```python
fig = px.imshow(
    token_embeddings_with_pe,
    color_continuous_scale="RdBu",
    labels=dict(x="Embedding Dimension", y="Token Index", color="Value"),
    title="Token Embedding with Positional Encoding Heatmap"
)

fig.update_xaxes(side="top")
fig.update_layout(height=500, width=900)
fig.show()
```

```python
te_with_pe_values = token_embeddings_with_pe.flatten()

fig_hist = px.histogram(te_with_pe_values, nbins=100, title='Distribution of Token Embedding Values with Positional Encoding')

fig_hist.update_layout(
    xaxis_title='Value',
    yaxis_title='Count',
    bargap=0.05,
    showlegend=False
)

fig_hist.show()
```
